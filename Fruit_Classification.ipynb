```python
# Import the matplotlib library's plotting module
# "plt" is a short name (alias) that we use instead of writing matplotlib.pyplot
import matplotlib.pyplot as plt


# Import the plot_tree function from sklearn.tree
# plot_tree() is used to DRAW/visualize a Decision Tree
from sklearn.tree import plot_tree


# Import the DecisionTreeClassifier class from sklearn.tree
# DecisionTreeClassifier is used to CREATE a Decision Tree model
# for classification problems
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [

    # 🍎 Apples
    # Weight = 145g, Color = 1 (Red), Texture = 1 (Smooth)
    [145, 1, 1],

    # Weight = 155g, Color = 1 (Red), Texture = 1 (Smooth)
    [155, 1, 1],

    # Weight = 165g, Color = 1 (Red), Texture = 2 (Rough)
    [165, 1, 2],

    # Weight = 175g, Color = 1 (Red), Texture = 1 (Smooth)
    [175, 1, 1],


    # 🍊 Oranges
    # Weight = 170g, Color = 2 (Orange), Texture = 2 (Rough)
    [170, 2, 2],

    # Weight = 185g, Color = 2 (Orange), Texture = 2 (Rough)
    [185, 2, 2],

    # Weight = 195g, Color = 2 (Orange), Texture = 2 (Rough)
    [195, 2, 2],

    # Weight = 205g, Color = 2 (Orange), Texture = 1 (Smooth)
    [205, 2, 1],


    # 🍌 Bananas
    # Weight = 105g, Color = 3 (Yellow), Texture = 1 (Smooth)
    [105, 3, 1],

    # Weight = 115g, Color = 3 (Yellow), Texture = 1 (Smooth)
    [115, 3, 1],

    # Weight = 125g, Color = 3 (Yellow), Texture = 1 (Smooth)
    [125, 3, 1],

    # Weight = 135g, Color = 3 (Yellow), Texture = 2 (Rough)
    [135, 3, 2],


    # 🥭 Mangoes
    # Weight = 210g, Color = 4 (Green/Yellow), Texture = 1 (Smooth)
    [210, 4, 1],

    # Weight = 230g, Color = 3 (Yellow), Texture = 1 (Smooth)
    [230, 3, 1],

    # Weight = 250g, Color = 4 (Green/Yellow), Texture = 1 (Smooth)
    [250, 4, 1]
]
```

```python
labels = [

    # 🍎 The first 4 rows in features are apples
    "apple",
    "apple",
    "apple",
    "apple",

    # 🍊 The next 4 rows are oranges
    "orange",
    "orange",
    "orange",
    "orange",

    # 🍌 The next 4 rows are bananas
    "banana",
    "banana",
    "banana",
    "banana",

    # 🥭 The last 3 rows are mangoes
    "mango",
    "mango",
    "mango"
]
```

```python
# Create a Decision Tree Classifier model
model = DecisionTreeClassifier()
```

```python
model.fit(features, labels)
```

```python
# Create some NEW fruits for testing the trained model
# These fruits were NOT used during training

test_fruits = [

    # [Weight, Color, Texture]
    # 150g, Red, Smooth
    # Expected answer: Apple 🍎
    [150, 1, 1],

    # 190g, Orange, Rough
    # Expected answer: Orange 🍊
    [190, 2, 2],

    # 110g, Yellow, Smooth
    # Expected answer: Banana 🍌
    [110, 3, 1],

    # 240g, Green/Yellow, Smooth
    # Expected answer: Mango 🥭
    [240, 4, 1]
]
```

```python
# Ask the trained model to predict the type of each test fruit
predictions = model.predict(test_fruits)
```

```python
# ============================================================
# PART 1: Display the predictions
# ============================================================


# Print a heading so we know that the following output
# contains our fruit classification predictions
print("--- Fruit Classification Predictions ---")


# Loop through every fruit in our test_fruits list
#
# "i" = number of the sample (1, 2, 3, 4...)
# "test_sample" = the actual fruit data, for example [150, 1, 1]
#
# enumerate(test_fruits, 1)
# means:
#   Start counting from 1 instead of 0
for i, test_sample in enumerate(test_fruits, 1):

    # Print:
    # Sample number
    # The fruit's features
    # The model's prediction
    #
    # predictions[i-1] is used because Python lists
    # start counting from 0.
    print(
        f"Sample {i} {test_sample} -> "
        f"Predicted Fruit: {predictions[i-1]}"
    )

```

```python

# ============================================================
# PART 2: Import tools for drawing the Decision Tree
# ============================================================


# Import matplotlib's plotting module.
# "plt" is a short name for matplotlib.pyplot.
# We use it to create and display graphs.
import matplotlib.pyplot as plt


# Import plot_tree.
# plot_tree() is used to visually draw our Decision Tree.
from sklearn.tree import plot_tree
from sklearn.tree import DecisionTreeClassifier
```

```python
# ============================================================
# PART 3: Create an area where we will draw the tree
# ============================================================


# Create a figure (drawing area) for our Decision Tree.
#
# figsize=(10, 6) means:
#   10 = width
#   6  = height
#
# The numbers are measured in inches.
plt.figure(figsize=(10, 6))


# ============================================================
# PART 4: Draw the Decision Tree
# ============================================================


plot_tree(
    model,

    # Give meaningful names to our three features.
    #
    # Feature 1 = Weight
    # Feature 2 = Color
    # Feature 3 = Texture
    feature_names=['Weight', 'Color', 'Texture'],

    # Show the names of the classes/fruits.
    #
    # model.classes_ contains:
    # ['apple', 'banana', 'mango', 'orange']
    #
    # These are the fruit names the model learned.
    class_names=model.classes_,

    # Color the boxes of the tree.
    # This makes the tree easier to understand visually.
    filled=True,

    # Make the corners of the boxes rounded.
    rounded=True
)


# ============================================================
# PART 5: Give the graph a title
# ============================================================


# Add a title above the Decision Tree.
plt.title("Fruit Classification Decision Tree")


# ============================================================
# PART 6: Display the Decision Tree
# ============================================================


# Actually show the graph on the screen.
plt.show()
```
